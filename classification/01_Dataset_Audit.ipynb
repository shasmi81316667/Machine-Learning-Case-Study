# 01 – Dataset Audit

**Dataset:** Steel Plate Faults

```python
import pandas as pd
from IPython.display import display

df = pd.read_csv("faults_with_missing(1).csv")

print("Shape:", df.shape)
print("\nFirst five rows:")
display(df.head())

print("\nData types:")
display(df.dtypes.to_frame("dtype"))

audit = pd.DataFrame({
    "dtype": df.dtypes.astype(str),
    "missing_count": df.isna().sum(),
    "missing_percent": (df.isna().mean()*100).round(2),
    "unique_values": df.nunique()
})
print("\nDataset audit:")
display(audit)

print("Duplicate rows:", int(df.duplicated().sum()))

TARGET_COLS = ["Pastry","Z_Scratch","K_Scatch","Stains","Dirtiness","Bumps","Other_Faults"]
print("\nTarget flag counts:")
display(df[TARGET_COLS].sum().sort_values(ascending=False).to_frame("count"))

target_sum = df[TARGET_COLS].sum(axis=1)
print("\nNumber of active target flags per row:")
display(target_sum.value_counts().sort_index().to_frame("rows"))

assert (target_sum == 1).all()
df["Fault_Type"] = df[TARGET_COLS].idxmax(axis=1)

print("\nFinal multiclass target distribution:")
display(df["Fault_Type"].value_counts().to_frame("count"))
```
